```python
import json
from pathlib import Path

folder_path = Path(".")
data_list = []

for file_path in folder_path.glob("*.json"):
    with file_path.open("r", encoding="utf-8") as f:
        data_list.append(json.load(f))

print(f"Loaded {len(data_list)} dictionaries.")
```

```python
benchmark1 = list(range(len(data_list)))
benchmark1 = sorted(benchmark1, key=lambda i: data_list[i]["benchmark1"]["best_metrics"]["test_pairwise_f1"], reverse=True)
for i in benchmark1:
    print(f"run {i+1}, {data_list[i]['benchmark1']['best_metrics']['test_pairwise_f1']}")
```

```python
benchmark2 = list(range(len(data_list)))
benchmark2 = sorted(benchmark2, key=lambda i: data_list[i]["benchmark2"]["best_metrics"]["test_pairwise_f1"], reverse=True)
for i in benchmark2:
    print(f"run {i+1}, {data_list[i]['benchmark2']['best_metrics']['test_pairwise_f1']}")
```

```python
average_test_f1 = []
for i in range(len(data_list)):
    sum = 0
    for benchmark in range(1,16):
        sum += data_list[i]['benchmark'+str(benchmark)]['best_metrics']['test_pairwise_f1']
    average_test_f1.append(sum/15)
average_test_f1
```

```python
avg = list(range(len(data_list)))
avg = sorted(avg, key=lambda i: average_test_f1[i], reverse=True)
for i in avg:
    print(f"run {i+1}, {average_test_f1[i]}")
```

```python
avg = list(range(len(data_list)))
avg = sorted(avg, key=lambda i: average_test_ari[i], reverse=True)
for i in avg:
    print(f"run {i+1}, {average_test_ari[i]}")
```
